Q1/10
Given the following sales data:
sales = np.array([[101, 250, 3],[102, 600, 5],[103, 450, 2],[104, 800, 6]])
Save the array to sales.txt, load it into data, and print the rows where the second column value is greater than 500.

```python
import numpy as np 
sales = np.array([[101, 250, 3],[102, 600, 5],[103, 450, 2],[104, 800, 6]])
np.savetxt("sales.txt",sales)
data = np.loadtxt("sales.txt")
print(data[data[:, 1] > 500])
```

Q6/10 — Broadcasting
NumPy broadcasting lets you perform operations between compatible array shapes without using loops.
A store tracks weekly sales of three products:
Calculate the revenue for each product in every week using NumPy, without using loops.

```python
sales = np.array([[100, 200, 150],
                  [120, 180, 160],
                  [90, 210, 170],
                  [140, 190, 180]])

prices = np.array([10, 20, 15])
revenue = prices * sales
revenue
```

Q10/10
A company tracks the quantities sold for three products across four weeks. Each row represents a week, and each column represents a product. Unit prices are ₹10, ₹20, and ₹15 for Products A, B, and C, respectively.
Calculate the revenue for each product in every week, then calculate the total revenue across all four weeks for each product.

```python
sales = np.array([[100, 200, 150],
                  [120, 180, 160],
                  [90, 210, 170],
                  [140, 190, 180]])

prices = np.array([10, 20, 15])

revenue = sales * prices
print(revenue)

total_revenue = revenue.sum(axis=0)
print(total_revenue)
```
