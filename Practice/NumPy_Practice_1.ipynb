Q1 — Array Creation & Properties 🟢

Create this NumPy array:

[10, 20, 30, 40, 50, 60]

Then print:

Number of dimensions
Shape
Number of elements
Data type
Item size

```python
import numpy as np
a = np.array([10, 20, 30, 40, 50, 60])
print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
print(a.itemsize)
```

Q2 — Indexing + Slicing

Create this array:

[10, 20, 30, 40, 50, 60, 70, 80]

Find:

First element
Last element
Elements from index 2 to 5
Last 3 elements
Elements at index 1, 3, and 5

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print(arr[0])
print(arr[-1])
print(arr[2:6])
print(arr[-3:])
print(arr[[1, 3, 5]])
```

Q3 — 2D Indexing + Slicing

Create:

[[10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]]

Find:

50
30
Last row
First column
First 2 rows and last 2 columns

```python
a = np.array([[10, 20, 30],
 [40, 50, 60],
 [70, 80, 90]])

print(a[1,[1]])
print(a[0,[2]])
print(a[-1])
print(a[:,0])
print(a[0:2,1:3])
```

Q5 — flatten() + Iteration

Given:

a = np.array([[1, 2, 3],
              [4, 5, 6]])

Use flatten() to convert it into a 1D array, then print each element one by one.

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

for i in a.flatten():
    print(i)
```

Q6 — np.nditer() Basic

Given:

a = np.array([[10, 20, 30],
              [40, 50, 60]])

Use np.nditer() to print each element one by one.

```python
a = np.array([[10, 20, 30],
              [40, 50, 60]])
for row in np.nditer(a):
    print(row)
```

Q7 — nditer() C Order vs F Order

Given the array:

a = np.array([[1, 2, 3],
              [4, 5, 6]])

Using np.nditer():

Print all elements in C order.
Print all elements in F order.

Write the code for both separately.

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

for i in np.nditer(a,order='C'):
    print(i)
```

```python
for i in np.nditer(a,order='F'):
    print(i)
```

Q8 — external_loop

Given:

a = np.arange(12).reshape(3, 4)

Use np.nditer() with external_loop and print the output.

```python
a = np.arange(12).reshape(3,4)
for x in np.nditer(a,flags=['external_loop']):
    print(x)
```

Q9 — readwrite

Given:

a = np.arange(1, 7).reshape(2, 3)

Use np.nditer() with op_flags=['readwrite'] and multiply every element by 2

```python
a = np.arange(1, 7).reshape(2, 3)
for x in np.nditer(a,op_flags=['readwrite']):
    x[...]=x*2
    print(a)
```
