Q15/15 — Final Lecture 1 Question
A company has 5 products with the following prices:
[100, 200, 300, 400, 500]
The company gives a 10% discount on every product.
Using NumPy:
1. Create the price array.
2. Calculate the discount amount for each product.
3. Calculate the final price after discount.
Hint: 10% = 0.10
Do it without a loop.

```python
price = np.array([100, 200, 300, 400, 500])
discount = price * 0.10
discount
```

```python
final_price = price - discount
final_price
```
