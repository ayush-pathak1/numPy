Q1 — Revision: axis
A company tracks the monthly sales of three products across four stores. Each row represents a store, and each column represents a product.
Calculate the total sales of each product across all four stores.

```python
import numpy as np
sales = np.array([[120, 150, 180],
                  [200, 170, 160],
                  [100, 140, 190],
                  [130, 160, 210]])

total_sales = sales.sum(axis=0)
print(total_sales)
```

Q2/10 — Revision: axis
Using the same sales array, calculate the total sales for each store across all three products.

```python
monthly_sales = sales.sum(axis=1)
print(monthly_sales)
```

Q3/10 — Broadcasting
A company sells three products across four stores. Each row represents a store, and each column represents a product. Unit prices are ₹20, ₹30, and ₹40, respectively.
Calculate the revenue for each product at every store.

```python
sales = np.array([[10, 20, 15],
                  [12, 18, 16],
                  [15, 10, 20],
                  [8,  25, 12]])

prices = np.array([20, 30, 40])
revenue =  sales*prices
print(revenue)
```

```python
total_revenue = revenue.sum(axis=1)
print(total_revenue)
```

Q5/10 — Revenue + axis

Q6/10 — Mixed Practice
A company records monthly sales quantities for three products. Each row represents a month, and each column represents a product.
Calculate the revenue for each product using the given prices, then use np.where() to label each month's total revenue as "Target Met" if it is at least ₹2,000, or "Below Target" otherwise.

```python
sales = np.array([[20, 30, 15],
                  [10, 25, 20],
                  [30, 20, 25]])

prices = np.array([40, 30, 50])
revenue = sales*prices
monthly_total = revenue.sum(axis=1)
label = np.where(monthly_total >= 2000,'Target Met', 'Below Target')
print(label)
```

Q7/10 — Boolean Indexing + Broadcasting
A company records the quantities sold for three products across three months. Each row represents a month, and each column represents a product. The unit prices are ₹25, ₹40, and ₹30, respectively.
Calculate the revenue for each product in every month, then display only the individual revenue values greater than ₹800.

```python
sales = np.array([[20, 25, 10],
                  [15, 30, 20],
                  [40, 10, 15]])

prices = np.array([25, 40, 30])
product_revenue = sales*prices
product_revenue[product_revenue > 800]
```

Q8/10
Using the same sales and prices arrays, calculate the revenue for each product in every month, then display the complete revenue rows for months whose total revenue is greater than ₹2,000.

```python
sales = np.array([[20, 25, 10],
                  [15, 30, 20],
                  [40, 10, 15]])

prices = np.array([25, 40, 30])
product_revenue = sales*prices
monthly_revenue = product_revenue.sum(axis=1)
total_greater_2000 = product_revenue[monthly_revenue > 2000]
print(total_greater_2000)
```

Q9/10
Using the product_revenue array, use np.where() to replace revenue values below ₹500 with 0, while keeping all other values unchanged. Display the modified array.

```python
replace_revenue = np.where(product_revenue < 500, 0,product_revenue)
replace_revenue
```

Q10/10 — Final Mixed Challenge
A company records sales quantities for three products across four months. Each row represents a month, and each column represents a product. Unit prices are ₹50, ₹30, and ₹40, respectively.
Calculate the revenue for each product in every month, calculate the total revenue for each month, and use np.where() to label each month as "Target Met" if its total revenue is at least ₹4,000, otherwise "Below Target".

```python
sales = np.array([[20, 30, 25],
                  [15, 40, 20],
                  [30, 25, 35],
                  [10, 50, 15]])

prices = np.array([50, 30, 40])
product_revenue_on_month = sales*prices
total_revenue = product_revenue_on_month.sum(axis=1)
lable = np.where(total_revenue >= 4000, 'Target Met', 'Below Target')
print(lable)
```
