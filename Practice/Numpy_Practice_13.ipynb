Q1/10 — np.argsort()
A company records monthly revenue for three products. Each row represents a product, and each column represents a month.
revenue = np.array([[1800, 1200, 2500, 1600],
                    [3000, 2200, 1800, 2800],
                    [1500, 2100, 1900, 2600]])
Use np.argsort() to find the month indices arranged from highest revenue to lowest revenue for each product. Store the result in ranked_months.


```python
import numpy as np
revenue = np.array([[1800, 1200, 2500, 1600],
                    [3000, 2200, 1800, 2800],
                    [1500, 2100, 1900, 2600]])

ranked_months = np.argsort(revenue,axis=1)[:,::-1]
ranked_months
```

Q2/10 — Independent Recall
Using the same revenue array, find the month index of the lowest revenue for each product.

```python
lowest_revenue = np.argmin(revenue,axis=1)
lowest_revenue
```

Q3/10 — np.unique() + Frequency
A company records customer purchase categories:
categories = np.array(['Electronics', 'Clothing', 'Electronics','Grocery', 'Clothing', 'Electronics',Grocery', 'Clothing'])

Find each unique category and its frequency. Store the results in two separate variables

```python
categories = np.array(['Electronics', 'Clothing', 'Electronics',
                       'Grocery', 'Clothing', 'Electronics',
                       'Grocery', 'Clothing'])
unique_categories, counts = np.unique(categories, return_counts=True)
print(unique_categories,counts)
```

Q4/10 — np.argmax() + Indexing
Using the unique_categories and counts arrays from Q3, find:
1. The category with the highest frequency.
2. Its frequency count.
Store them in separate variables.

```python
idx = np.argmax(counts)
highest_freq = unique_categories[idx]
frequency_count = counts[idx]
print(highest_freq,frequency_count)
```

Q5/10 — NaN Handling
A company records monthly sales for three products. Each row represents a month, and each column represents a product.
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350],
                  [np.nan, 300, 400]])
1. Count the missing values in each product column.
2. Calculate the average sales for each product, ignoring missing values.
Store both results in separate variables.

```python
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350],
                  [np.nan, 300, 400]])

count_missing = np.isnan(sales).sum(axis=0)
average_sales = np.nanmean(sales, axis=0)
print(count_missing,average_sales)
```

Q6/10 — Handling Missing Values
A company records monthly sales for three products. Some values are missing.
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350]])
Use np.where() with np.isnan() to replace all missing values with 0, while keeping the other values unchanged. Store the result in a new array named clean_sales.

```python
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350]])
clean_sales = np.where(np.isnan(sales),0,sales)
print(clean_sales)
```

Q7/10 — Frequency Analysis
A retail company records product IDs from customer transactions.
product_ids = np.array([205, 201, 203, 205, 202, 201,
                        205, 203, 201, 205, 204, 201])
Find the most frequently purchased product ID and its total number of transactions. Store both results in separate variables.

```python
product_ids = np.array([205, 201, 203, 205, 202, 201,
                        205, 203, 201, 205, 204, 201])
freq,counts= np.unique(product_ids,return_counts=True)
idx = np.argmax(counts)
frequent_product_id = freq[idx]
total_number_trans = counts[idx]
print(frequent_product_id,total_number_trans)
```

Q8/10 — Frequency Analysis (Advanced)
Using the same product_ids array, identify all product IDs that share the highest transaction frequency. Store all matching product IDs in a new array named top_products.

```python
product_id, count_freq = np.unique(product_ids, return_counts=True)
max_freq = np.max(count_freq)
top_products = product_id[count_freq == max_freq]
top_products
```
