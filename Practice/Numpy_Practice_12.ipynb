Q1/10 — Quick Recall
A company tracks sales of three products across four months. Each row represents a month, and each column represents a product.

Calculate the revenue for each product in every month, then calculate the total revenue for each month.

```python
import numpy as np
sales = np.array([[20, 30, 25],
                  [15, 40, 20],
                  [30, 25, 35],
                  [10, 50, 15]])

prices = np.array([50, 30, 40])
product_revenue = sales * prices
total_revenue = product_revenue.sum(axis=1)
print(product_revenue)
print(total_revenue)
```

Q2/10 — np.where() Revision
Using the total_revenue array, label each month:
- Revenue of ₹3,000 or more → "Target Met"
- Revenue below ₹3,000 → "Below Target"


```python
np.where(total_revenue >= 3000, "Target Met", "Below Target")
```

Q3/10 — np.argmax()
Using the product_revenue array from Q1, find the month index (0-based) in which each product achieved its highest revenue.

Store the three indices in a new variable named best_month.

```python
best_month = np.argmax(product_revenue,axis=0)
print(best_month)
```

Q4/10 — np.sort() Practical
Using the product_revenue array from Q1, sort the monthly revenue values of each product in ascending order. Store the result in a new array without modifying the original array.

```python
sorted_revenue = np.sort(product_revenue,axis=0)
print(sorted_revenue)
```

Q5/10 — np.argsort()
Using product_revenue, find the month indices arranged from highest revenue to lowest revenue for each product.

Store the result in ranked_months.

```python
ranked_months = np.argsort(product_revenue,axis=0)[::-1]
print(ranked_months)
```

Q7/10 — np.unique() with Frequency Counting
A company records the product IDs sold in different transactions.
product_ids = np.array([101, 102, 101, 103, 102, 101, 104, 103, 102, 101])
Find each unique product ID and count how many times it appears. Store the results in two separate variables.

```python
product_ids = np.array([101, 102, 101, 103, 102, 101, 104, 103, 102, 101])
unique_id,counts = np.unique(product_ids,return_counts=True)
print(unique_id,counts)
```

Q8/10 — Frequency Analysis
Using the same product_ids array, find the product ID with the highest number of transactions and its transaction count.

```python
highest_transaction,count = np.unique(product_ids,return_counts=True)
idx = np.argmax(count)
highest_transaction = highest_transaction[idx]
count = count[idx]
print(highest_transaction)
print(count)
```

Q9/10 — Handling Missing Values (NaN)
New concept: np.isnan() checks which array elements are missing (NaN).

Given:
sales = np.array([1200, 1500, np.nan, 1800, np.nan, 900])
Find the number of missing values in the array and calculate the average of the available sales values, ignoring missing values.

```python
sales = np.array([1200, 1500, np.nan, 1800, np.nan, 900])
missing_count = np.isnan(sales).sum()
average_sales = np.nanmean(sales)
print(missing_count)
print(average_sales)
```

Q10/10 — Final Challenge: NaN + Statistics
A company tracks monthly sales of three products. Each row represents a month, and each column represents a product. Some sales values are missing.
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350],
                  [np.nan, 300, 400]])
                
1. Count the missing values in each product column.
2. Calculate the average sales for each product, ignoring missing values.
Store both results in separate variables.

```python
sales = np.array([[100, 200, np.nan],
                  [150, np.nan, 300],
                  [200, 250, 350],
                  [np.nan, 300, 400]])

missing_count = np.isnan(sales).sum(axis=0)
print(missing_count)
average_sales = np.nanmean(sales,axis=0)
print(average_sales)
```
