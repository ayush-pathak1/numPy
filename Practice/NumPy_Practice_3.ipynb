Lecture 2 — Q1
Create this array:
[[10, 20, 30],
 [40, 50, 60]]

Then print:
1. Number of dimensions
2. Shape
3. Number of elements
4. Data type
5. Size of each element in bytes

```python
import numpy as np
a = np.array([[10, 20, 30],
 [40, 50, 60]])

print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
print(a.itemsize)
```
