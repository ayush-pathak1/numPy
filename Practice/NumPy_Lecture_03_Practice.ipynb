Q13. Monthly sales:

sales = np.array([12000, 18000, 9000, 25000, 15000, 30000])

Q14. Employee salaries:

salary = np.array([25000, 40000, 30000, 55000, 35000])

Select salaries greater than 30,000.

Then replace those selected salaries with -1.

```python
salary = np.array([25000, 40000, 30000, 55000, 35000])
salary[salary > 30000]
```

```python
salary[salary > 30000] = -1
print(salary)
```
