Q1/10 — Frequency Analysis + Boolean Masking
An online store records product IDs for customer purchases.
product_ids = np.array([
    401, 402, 403, 401, 404,
    402, 405, 401, 402, 405,
    401, 402, 405, 404, 403
])
Task: Find all product IDs that have the highest purchase frequency. There may be a tie.

```python
import numpy as np
product_ids = np.array([
    401, 402, 403, 401, 404,
    402, 405, 401, 402, 405,
    401, 402, 405, 404, 403
])
unique_id,freq = np.unique(product_ids,return_counts=True)
max_freq = np.max(freq)
highest_purchase = freq == max_freq
print(unique_id[highest_purchase])
```

sales = np.array([

product_ids = np.array([101, 102, 103, 104])

Q3/10 — np.argsort() in 2D Array
A company tracks monthly revenue for 3 products.
revenue = np.array([
    [400, 200, 500, 300],
    [700, 900, 600, 800],
    [250, 450, 350, 550]
])
Task: Find the month indices in descending order of revenue for each product.

```python
revenue = np.array([
    [400, 200, 500, 300],
    [700, 900, 600, 800],
    [250, 450, 350, 550]
])
descending = np.argsort(revenue)[:,::-1]
print(descending)
```

Q5/10 — np.sort() vs .sort()
A company records employee salaries:
salary = np.array([35000, 25000, 45000, 30000, 40000])
Tasks:
1. Create a descending sorted copy using np.sort().
2. Print the original salary array without modifying it.
3. Sort the original array in ascending order using .sort().
4. Print the modified original array.

```python
salary = np.array([35000, 25000, 45000, 30000, 40000])
descending_sorted_copy = np.sort(salary)[::-1]
print("Descending:", descending_sorted_copy)
print("Original:", salary)
salary.sort()
print("Ascending:", salary)
```

Next Q6/10 — np.where() + Business Logic
A company tracks employee performance scores:
scores = np.array([45, 78, 90, 55, 82, 35])
Task: Create a result array:
- Score greater than or equal to 60 → "Pass"
- Score below 60 → "Fail"

```python
scores = np.array([45, 78, 90, 55, 82, 35])
result = np.where(scores >= 60, "Pass","Fail")
print(result)
```

Q10/10 — Final Business Challenge
A retail company tracks weekly revenue:
revenue = np.array([400, 500, 450, 600, 700, 650, 800])
Tasks:
1. Calculate the total weekly revenue using np.sum().
2. Calculate cumulative revenue after each day using np.cumsum().
3. Calculate daily revenue changes using np.diff().
4. Find the median daily revenue using np.median().

```python
revenue = np.array([400, 500, 450, 600, 700, 650, 800])
weekly_revenue = np.sum(revenue)
print("Weekly Revenue :",weekly_revenue)
cumulative_revenue = np.cumsum(revenue)
print("Cumulative Revenue :", cumulative_revenue)
revenue_changes = np.diff(revenue)
print("Revenue Changes :", revenue_changes)
median_revenue = np.median(revenue)
print("Median Revenue :", median_revenue)
```
