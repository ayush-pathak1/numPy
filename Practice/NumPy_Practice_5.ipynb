Q1 — Indexing & Slicing
Given:
a = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
Write NumPy code to print:
1. The element 50
2. The last row
3. The first 2 rows and last 2 columns
Use indexing/slicing only.

```python
import numpy as np
a = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
print(a[1,1])
print(a[-1,:])
print(a[0:2,-2:])
```

Q2 — Boolean Indexing
a = np.arange(1, 13).reshape(3, 4)
Write code to:
1. Select all even numbers
2. Replace all even numbers with -1

```python
a = np.arange(1, 13).reshape(3, 4)
a[a % 2 == 0]
```

```python
a[a % 2 == 0] = -1
a
```
