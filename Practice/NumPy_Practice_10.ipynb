Q1 — Revision: axis
A company tracks revenue from three products over four months. Each row represents a month, and each column represents a product.
Calculate the total revenue for each product across all four months.
revenue = np.array([[1200, 1500, 1800],
                    [2000, 1700, 1600],
                    [1000, 1400, 1900],
                    [1300, 1600, 2100]])

```python
import numpy as np

revenue = np.array([[1200, 1500, 1800],
                    [2000, 1700, 1600],
                    [1000, 1400, 1900],
                    [1300, 1600, 2100]])
total_revenue = revenue.sum(axis=0)
print(total_revenue)
```

Q2/10 — Revision
Using the same revenue array, calculate the total revenue for each month across all three products.

```python
total_revenue = revenue.sum(axis=1)
print(total_revenue)
```

Q3/10 — Broadcasting Revision
A store sells three products over three months. Each row represents a month, and each column represents a product. Unit prices are ₹20, ₹30, and ₹50, respectively.
Calculate the revenue for each product in every month.

```python
sales = np.array([[10, 20, 15],
                  [12, 18, 16],
                  [15, 10, 20]])

prices = np.array([20, 30, 50])

revenue = sales * prices
print(revenue)
```

Q4/10 — Broadcasting + axis
Using the revenue array from Q3, calculate the total revenue for each month.

```python
monthly_revenue = revenue.sum(axis=1)
print(monthly_revenue)
```

Q5/10 — Broadcasting + axis
Using the same revenue array from Q3, calculate the total revenue for each product across all three months

```python
product_revenue = revenue.sum(axis=0)
print(product_revenue)
```

Q6/10 — np.where() Revision
Given the monthly sales totals below, use np.where() to label totals of ₹2,000 or more as "Target Met" and all other totals as "Below Target".

```python
monthly_sales = np.array([2500, 1800, 2000, 1500, 3000])
label = np.where(monthly_sales >= 2000,'Target Met','Below Target')
print(label)
```

Q7/10 — Mixed Practice
A company tracks the quantities of three products sold across three months. Each row represents a month, and each column represents a product. The unit prices are ₹20, ₹30, and ₹40, respectively.
Calculate the revenue for each product in every month, then use np.where() to replace individual revenue values below ₹500 with 0, keeping all other values unchanged.

```python
sales = np.array([[10, 20, 15],
                  [25, 10, 20],
                  [15, 30, 10]])

prices = np.array([20, 30, 40])
revenue = sales * prices
label = np.where(revenue < 500, 0,revenue)
print(label)
```

Q8/10 — Mixed Practice
A company records sales quantities for three products across four months. Each row represents a month, and each column represents a product. Unit prices are ₹50, ₹30, and ₹40, respectively.
Calculate the revenue for each product in every month, then display the complete revenue rows for months whose total revenue is at least ₹3,000.

```python
sales = np.array([[20, 30, 25],
                  [15, 40, 20],
                  [30, 25, 35],
                  [10, 50, 15]])

prices = np.array([50, 30, 40])
revenue = sales * prices
monthly_revenue = revenue.sum(axis=1)
revenue[monthly_revenue >= 3000]
```

Q9/10 — Mixed Practice
A company records the quantities sold for three products across three months. Each row represents a month, and each column represents a product. The unit prices are ₹50, ₹30, and ₹40, respectively.
Calculate the revenue for each product in every month, replace individual revenue values below ₹1,000 with 0 using np.where(), then calculate the total adjusted revenue for each month.

```python
sales = np.array([[20, 30, 25],
                  [15, 40, 20],
                  [30, 25, 35]])

prices = np.array([50, 30, 40])
revenue = sales*prices
replace = np.where(revenue < 1000,0,revenue)
adjusted_revenue = replace.sum(axis=1)
print(adjusted_revenue)
```

Q10/10 — Final Challenge
A company tracks the quantities of three products sold across four months. Each row represents a month, and each column represents a product. Unit prices are ₹40, ₹30, and ₹50, respectively.
Calculate the revenue for each product in every month, calculate the total revenue for each month, and use np.where() to label each month as "Target Met" if its total revenue is at least ₹3,000, otherwise "Below Target".

```python
sales = np.array([[20, 25, 15],
                  [10, 30, 20],
                  [25, 20, 30],
                  [15, 35, 10]])

prices = np.array([40, 30, 50])
revenue = sales*prices
monthly_revenue = revenue.sum(axis=1)
label = np.where(monthly_revenue >= 3000,'Target Met','Below Target')
print(label)
```
