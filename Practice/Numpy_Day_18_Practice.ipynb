```python
import numpy as np

product_ids = np.array([101, 102, 103, 104, 105, 106])

monthly_sales = np.array([
    [120, 150, 180, 200, 220, 250],
    [90, 110,  np.nan, 140, 160, 180],
    [200, 190, 210, 230, 250, 270],
    [80,  95,  100,  np.nan, 130, 150],
    [150, 160, 170, 180, 190, 200],
    [110, 130, 150, 170, 190, 210]
], dtype=float)

prices = np.array([500, 750, 400, 900, 600, 350])

customer_ids = np.array([
    201, 202, 203, 201, 204, 202, 205, 201,
    203, 206, 202, 204, 201, 205, 203, 202,
    207, 201, 204, 203
])

daily_revenue = np.array([
    1200, 1500, 1100, 1800, 2000,
    1700, 2200, 1900, 2500, 2100
])
```

Question 1/7 — Cumulative revenue
The company wants to monitor its running revenue over 10 days.
Task:
1. Calculate the cumulative revenue for each day.
2. Find the total revenue generated over all 10 days.

```python
cumulative_revenue = np.cumsum(daily_revenue)
print(cumulative_revenue)
total_revenue = np.sum(daily_revenue)
print(total_revenue)
```

Question 2/7 — Daily revenue changes
Management wants to understand how revenue changes from one day to the next.
Tasks:
1. Calculate the day-to-day revenue difference.
2. Identify the largest positive increase.
3. Identify the largest negative change (drop).

```python
revenue_difference = np.diff(daily_revenue)
print(revenue_difference)
largest_positive = np.max(revenue_difference)
print(largest_positive)
largest_negative = np.min(revenue_difference)
print(largest_negative)
```

Question 3/7 — Customer purchase frequency
The company wants to identify its most frequent customers.
Using the existing customer_ids array:
1. Find all unique customer IDs and their purchase counts.
2. Find the maximum purchase frequency.
3. Identify all customers who have that maximum frequency (There may be a tie).

```python
unique_id,purchase_counts = np.unique(customer_ids,return_counts=True)
max_freq = np.max(purchase_counts)
mask = purchase_counts == max_freq
max_freq_ids = unique_id[mask]
print(max_freq_ids)
```

Question 4/7 — Product-wise sorting
The company wants to rank each product's monthly sales from highest to lowest.
Using monthly_sales:
1. Sort the monthly sales of each product in descending order.
2. Keep the original monthly_sales array unchanged.
3. Handle the missing values (NaN) appropriately so they don't appear at the beginning of the ranking.

```python
clean_sales = np.nan_to_num(monthly_sales, nan=-np.inf)
sorted_sales = np.sort(clean_sales,axis=1)[:,::-1]
print(sorted_sales)
```

Question 5/7 — Missing sales analysis
The management team wants to evaluate data quality and average performance.
Using monthly_sales:
1. Count the missing sales values for each product.
2. Calculate the average monthly sales for each product, ignoring missing values.
Condition: Do not replace or modify the original NaN values.

```python
count_missing_sales = np.isnan(monthly_sales).sum(axis=1)
print(count_missing_sales)
average_sales = np.nanmean(monthly_sales,axis=1)
print(average_sales)
```

Question 6/7 — Revenue target analysis
The company has a monthly revenue target of ₹1,500.
Using daily_revenue:
1. Classify each day as "Target Achieved" if revenue is greater than or equal to 1500.
2. Otherwise classify it as "Target Not Achieved".
3. Count the total number of days when the target was achieved.
Use NumPy.

```python
classify_days = np.where(daily_revenue >= 1500,"Target Achieved","Target Not Achieved")
print(classify_days)
target_achieved_count = np.sum(daily_revenue >= 1500)
print(target_achieved_count)
```

Question 7/7 — Product-wise revenue
The company wants to calculate total revenue generated by each product.
Use:
- monthly_sales
- prices
- product_ids
Tasks:
1. Replace missing sales values with 0.
2. Calculate total units sold by each product across all six months.
3. Calculate total revenue for each product (units sold × product price).
4. Identify the product ID with the highest total revenue.
Try independently. This question tests NaN handling, axis selection, broadcasting, and argmax().

```python
missing_mask = np.isnan(monthly_sales)
clean_sales = np.where(missing_mask, 0, monthly_sales)
units_sold = np.sum(clean_sales, axis=1)
product_revenue = units_sold * prices
highest_revenue_idx = np.argmax(product_revenue)
top_product_id = product_ids[highest_revenue_idx]
top_product_revenue = product_revenue[highest_revenue_idx]
print("Units Sold:", units_sold)
print("Product Revenue:", product_revenue)
print("Top Product ID:", top_product_id)
print("Highest Revenue:", top_product_revenue)
```
