Q1/10 — Boolean Masking + Frequency Analysis
An e-commerce company records the following product IDs for customer transactions.
product_ids = np.array([
    301, 302, 303, 301, 304,
    302, 305, 301, 302, 305,
    301, 302, 305, 304
])
Task: Find all product IDs that have the highest transaction frequency.

```python
import  numpy as np
product_ids = np.array([
    301, 302, 303, 301, 304,
    302, 305, 301, 302, 305,
    301, 302, 305, 304
])
unique_id,freq = np.unique(product_ids,return_counts=True)
max_freq = np.max(freq)
max_freq = freq == max_freq
print(unique_id[max_freq])
```

Q4/10 — np.argsort() + Descending Order
An e-commerce company records revenue for 5 products.
revenue = np.array([750, 900, 600, 850, 500])
product_ids = np.array([401, 402, 403, 404, 405])
Task: Find the product IDs in descending order of revenue.

```python
revenue = np.array([750, 900, 600, 850, 500])
product_ids = np.array([401, 402, 403, 404, 405])
sorted_revenue = np.argsort(revenue)[::-1]
print(product_ids[sorted_revenue])
```

Q5/10 — np.argmax() + Index Retrieval
A company tracks revenue generated by different products.
revenue = np.array([450, 800, 650, 950, 700])
product_ids = np.array([501, 502, 503, 504, 505])
Task: Find the product ID with the highest revenue and its revenue value.

```python
revenue = np.array([450, 800, 650, 950, 700])
product_ids = np.array([501, 502, 503, 504, 505])
highest_revenue = np.argmax(revenue)
product_id = product_ids[highest_revenue]
print("Product ID:",product_id)
print("Revenue:",revenue[highest_revenue])
```

Q6/10 — Boolean Masking + Conditional Selection
A company tracks employee performance scores.
scores = np.array([65, 85, 90, 85, 72, 90, 60, 90])
employee_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108])
Task: Find all employee IDs whose score is equal to the highest score.

```python
scores = np.array([65, 85, 90, 85, 72, 90, 60, 90])
employee_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108])
highest_score = np.max(scores)
mask = scores == highest_score
highest_score_ids = employee_ids[mask]
print(highest_score_ids)
```

Q10/10 — Final Business Challenge
A retail company tracks quarterly sales and targets for three products.
sales = np.array([
    [120, 150, 180],
    [200, 170, 230],
    [140, 190, 160]
])

product_ids = np.array([101, 102, 103])
targets = np.array([400, 550, 500])
Tasks:
1. Calculate total quarterly sales for each product.
2. Compare each product's total sales with its target.
3. Display "Achieved" if sales are greater than or equal to the target; otherwise display "Not Achieved".
4. Display the product IDs with their respective status.
Try independently. This combines axis, comparison and np.where(). 💪

```python
sales = np.array([
    [120, 150, 180],
    [200, 170, 230],
    [140, 190, 160]
])

product_ids = np.array([101, 102, 103])
targets = np.array([400, 550, 500])
Qr_sales = sales.sum(axis=1)
comparision = np.where(Qr_sales >= targets,"Achieved","Not Achieved")
print(comparision)
for i, result in zip(product_ids,comparision):
    print(i,":",result)
```
