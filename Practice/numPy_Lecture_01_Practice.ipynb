### **Q8. Python List vs NumPy Speed**

Create two Python lists of $1,000,000$ elements each and measure the execution time for their element-wise addition. Then perform the exact same operation using two NumPy arrays and compare the execution times.

> **Hint:** Use `time.perf_counter()` or `%timeit` in Jupyter Notebook.

```python
size = 1000000

l1 = list(range(size))
l2 = list(range(size))
start = time.perf_counter()
result = [x + y for x, y in zip(l1, l2)]
python_time = time.perf_counter() - start
print("Python list took:", python_time * 1000, "ms")
```

```python
a1 = np.arange(size)
a2 = np.arange(size)
start = time.perf_counter()
result = a1 + a2
numpy_time = time.perf_counter() - start
print("NumPy took:", numpy_time * 1000, "ms")
```
