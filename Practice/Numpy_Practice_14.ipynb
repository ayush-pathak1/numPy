Q1/10 — Missing Values (Pending Q9)
A sales array contains monthly sales for 4 products. Each column represents one product, and NaN represents missing data.
sales = np.array([
    [120, np.nan, 300, 450],
    [150, 200, np.nan, 500],
    [np.nan, 250, 350, 550],
    [180, 300, 400, np.nan]
])
Task: Count the number of missing values in each product column.

```python
import numpy as np
sales = np.array([
    [120, np.nan, 300, 450],
    [150, 200, np.nan, 500],
    [np.nan, 250, 350, 550],
    [180, 300, 400, np.nan]
])
count_misssing = np.isnan(sales).sum(axis=0)
print(count_misssing)
```

Q3/10 — Frequency Analysis
A retail company recorded product IDs for its transactions:
product_ids = np.array([
    101, 102, 101, 103, 102,
    101, 104, 103, 102, 101
])
Task: Find:
1. All unique product IDs.
2. The transaction count for each product ID.

```python
product_ids = np.array([
    101, 102, 101, 103, 102,
    101, 104, 103, 102, 101
])
unique_id, transaction_count = np.unique(product_ids,return_counts=True)
print(unique_id, transaction_count)
```

Highest Frequency ProductQ4/10 — 
Using the same unique_id and transaction_count arrays:

Task: Find the product ID with the highest number of transactions and its transaction count

```python
idx = np.argmax(transaction_count)
highest_product = unique_id[idx]
highest_count = transaction_count[idx]
print(highest_product,highest_count)
```

Q5/10 — Find All Top-Selling Products
A company wants to identify every product tied for the highest transaction frequency.
product_ids = np.array([
    201, 202, 203, 201, 204,
    202, 205, 201, 205, 202,
    205, 204
])
Task: Find all product IDs having the maximum transaction count.
There may be a tie. Return all matching product IDs, not just one.

```python
ids, count = np.unique(product_ids, return_counts=True)
max_count = np.max(count)
tie_ids = count == max_count
all_ids = ids[tie_ids]
print(all_ids)
```

Q6/10 — Replace Missing Values
A company has recorded incomplete sales data:
sales = np.array([120, np.nan, 250, np.nan, 400, 180])
Task: Replace every NaN value with 0 using np.where().

```python
sales = np.array([120, np.nan, 250, np.nan, 400, 180])
replaced = np.where(np.isnan(sales),0,sales)
print(replaced)
```

Q7/10 — Sorting vs In-place Sorting
A company stores monthly revenue:
revenue = np.array([450, 200, 600, 300, 500])
Task:
1. Sort revenue in ascending order using np.sort() without changing the original array.
2. Then sort the original array in descending order using the in-place .sort() method.
Print both results and the original array after step 1 (before in-place sorting, if needed).

```python
revenue = np.array([450, 200, 600, 300, 500])
sorted_revenue = np.sort(revenue)
print(sorted_revenue)
print(revenue)

revenue.sort()
descending_revenue = revenue[::-1]
print(descending_revenue)
```

Q8/10 — Find Maximum Index
sales = np.array([
    [120, 250, 180],
    [300, 150, 400],
    [220, 350, 280]
])
Each row represents a product, and each column represents a month.
Task: Find the column index (month index) of the highest sales for each product.

```python
sales = np.array([
    [120, 250, 180],
    [300, 150, 400],
    [220, 350, 280]
])
monthly_revenue = sales.sum(axis=1)
max_idx = np.argmax(monthly_revenue)
print(max_idx)
```

Q9/10 — np.argsort() (Descending Order)
A company tracks sales of 5 products:
sales = np.array([450, 200, 600, 300, 500])
product_ids = np.array([101, 102, 103, 104, 105])
Task: Find the product IDs in descending order of sales

```python
sorted_indices = np.argsort(sales)[::-1]
sorted_product_ids = product_ids[sorted_indices]
print(sorted_product_ids)
```

Q10/10 — Business Analysis Challenge
A company records sales of 4 products across 3 months.
sales = np.array([
    [120, 250, 180],
    [300, np.nan, 400],
    [220, 350, 280],
    [np.nan, 150, 200]
])

product_ids = np.array([101, 102, 103, 104])
Tasks:
1. Calculate average sales for each product, ignoring NaN.
2. Find the product ID with the highest average sales.

```python
average_sale = np.nanmean(sales, axis=1)

highest_average_sale = np.argmax(average_sale)
product_id = product_ids[highest_average_sale]

print("Average Sales:", average_sale)
print("Product ID:", product_id)
```
