Q20. ⭐ Challenge

price = np.array([100, 200, 300, 400])
quantity = np.array([2, 1, 3, 2])

Find:

Total amount for each product
Overall total revenue
Average product revenue

```python
price = np.array([100, 200, 300, 400])
quantity = np.array([2, 1, 3, 2])
```

```python
product_amounts  = price * quantity
product_amounts
```

```python
total_revenue = price.sum()
total_revenue
```

```python
avg_revenue = product_amounts.mean()
avg_revenue
```
