Q1/10 — Recall Challenge
A company records sales quantities for three products across four months. Each row represents a month, and each column represents a product.

Calculate the revenue for each product in every month, then display the complete rows for months whose total revenue is greater than ₹3,000.

```python
import numpy as np

sales = np.array([[20, 30, 25],
                  [15, 40, 20],
                  [30, 25, 35],
                  [10, 50, 15]])

prices = np.array([50, 30, 40])
product_rev = sales * prices
total_rev  = product_rev.sum(axis=1)
product_rev[total_rev > 3000]
```

Q2/10 — Sorting
A company records monthly revenue for three products. Each row represents a product, and each column represents a month
revenue = np.array([[1200, 800, 1500, 1100],
                    [900, 1400, 1000, 1600],
                    [2000, 1800, 2200, 1700]])
Sort each product's monthly revenue in ascending order and store the result in a new array without modifying the original revenue array.

```python
revenue = np.array([[1200, 800, 1500, 1100],
                    [900, 1400, 1000, 1600],
                    [2000, 1800, 2200, 1700]])
sorted_monthly_revenue = np.sort(revenue,axis=1)
sorted_monthly_revenue
```

Q3/10 — Finding Maximum Position
New concept: np.argmax() returns the index position of the maximum value.

Given the array below, each row represents a product and each column represents a month.
revenue = np.array([[1200, 800, 1500, 1100],
                    [900, 1400, 1000, 1600],
                    [2000, 1800, 2200, 1700]])
Use NumPy to find the month index of the highest revenue for each product.

```python
revenue = np.array([[1200, 800, 1500, 1100],
                    [900, 1400, 1000, 1600],
                    [2000, 1800, 2200, 1700]])
np.argmax(revenue,axis=1)
```

Q4/10 — Sorting + Filtering
Given the array below, find the top 2 revenue values in each product's row, in descending order. Store the result in a new array and keep the original array unchanged.

```python
revenue = np.array([[1200, 800, 1500, 1100],
                    [900, 1400, 1000, 1600],
                    [2000, 1800, 2200, 1700]])
sorted_rev = np.sort(revenue, axis=1)
top_2 = sorted_rev[:, -2:][:, ::-1]
print(top_2)
```
