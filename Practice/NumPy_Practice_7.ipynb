Q3/10 — Broadcasting

Calculate the revenue for each product on each day, without using loops.

```python
sales = np.array([[10, 20, 15],
                  [12, 18, 16],
                  [9, 21, 17]])

prices = np.array([10, 20, 15])
revenue = sales * prices
revenue
```

Q8/10
A company tracks the quantities sold for three products across three months. Each row represents a month, and each column represents a product. Unit prices are ₹20, ₹30, and ₹10, respectively.
Calculate the revenue for each product in every month, then display the complete rows for months whose total revenue is greater than ₹1,000.

```python
sales = np.array([[10, 20, 30],
                  [20, 15, 10],
                  [15, 25, 20]])

prices = np.array([20, 30, 10])

revenue = sales * prices

monthly_total = revenue.sum(axis=1)

filtered_revenue = revenue[monthly_total > 1000]

print(filtered_revenue)
```

Q9/10 — Mixed Practice
A company tracks the quantities of three products sold each month. Each row represents a month, and each column represents a product. The unit prices are ₹30, ₹50, and ₹20 for Products A, B, and C, respectively.
Calculate the total revenue for each month, then display only the monthly totals greater than ₹1,000.

```python
sales = np.array([[8, 12, 15],
                  [5, 5, 10],
                  [10, 15, 20]])

prices = np.array([30, 50, 20])
revenue = sales * prices
total_revenue = revenue.sum(axis=1)
total_revenue[total_revenue > 1000]
```

Q10/10 — Final Challenge
A company tracks the quantities of three products sold over three months. Each row represents a month, and each column represents a product. Unit prices are ₹40, ₹25, and ₹30 for Products A, B, and C, respectively.
Calculate the total revenue for each product across all three months, then display only the product totals greater than ₹2,000.

```python
sales = np.array([[20, 30, 15],
                  [25, 40, 20],
                  [30, 35, 25]])

prices = np.array([40, 25, 30])

revenue = sales * prices
total_revenue = revenue.sum(axis=0)
total_revenue[total_revenue > 2000]
```
