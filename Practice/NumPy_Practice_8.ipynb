Q1/10 — Revision
A company records the sales quantities of three products across four weeks. Each row represents a week, and each column represents a product.
Calculate the total quantity sold for each product across all four weeks.

```python
import numpy as np

sales = np.array([[20, 35, 15],
                  [25, 30, 20],
                  [15, 40, 25],
                  [30, 25, 10]])

sales.sum(axis=0)
```

Using the same sales array, calculate the total quantity sold across all three products for each week.

Q3/10 — Broadcasting
A company sells three products with unit prices of ₹40, ₹25, and ₹30. Each row represents a week, and each column represents a product.
Calculate the revenue for each product in every week.

```python
sales = np.array([[10, 20, 15],
                  [12, 18, 16],
                  [15, 10, 20]])

prices = np.array([40, 25, 30])

revenue = sales * prices
revenue
```

Q4/10
Using the revenue array from Q3, calculate the total revenue for each week.

```python
revenue.sum(axis=1)
```

Q5/10 — Broadcasting + axis
Using the same revenue array from Q3, calculate the total revenue for each product across all three weeks.

```python
revenue.sum(axis=0)
```

Q6/10
Using the revenue array from Q3, select and display all individual revenue values greater than ₹500.

```python
revenue[revenue > 500]
```

Q7/10
Using the revenue array from Q3, replace all revenue values below ₹450 with 0 and display the modified array.

```python
revenue[revenue < 450] = 0
revenue
```

Q8/10 — np.where()
np.where(condition, value_if_true, value_if_false) lets you choose values based on a condition
Use np.where() to create a new array where values greater than or equal to 500 become "High" and all other values become "Low".

```python
sales = np.array([250, 600, 400, 800, 150])
result = np.where(sales >= 500,'High','low')
print(result)
```

Q9/10
Given the sales array, use np.where() to replace values below 300 with 0, while keeping all other values unchanged.

```python
sales = np.array([180, 420, 250, 600, 90])
result = np.where(sales < 300,0,sales)
result
```

Q10/10 — Final Challenge
A company tracks sales quantities for three products across three months. Each row represents a month.
Calculate the total quantity sold in each month, then use np.where() to label months with total sales of 100 or more as "Target Met" and the remaining months as "Below Target".

```python
sales = np.array([[40, 35, 30],
                  [20, 25, 15],
                  [50, 30, 25]])

result = sales.sum(axis=1)
monthly_status = np.where(result >= 100,'Target Met','Below Target')
print(monthly_status)
```
