Q1/10 — Axis Selection + np.argmax()
A company tracks monthly revenue for 4 products.
revenue = np.array([
    [120, 250, 180],
    [300, 150, 400],
    [220, 350, 280],
    [500, 450, 600]
])

product_ids = np.array([101, 102, 103, 104])
Task: Find the month index of the highest revenue for each product.

```python
import numpy as np
revenue = np.array([
    [120, 250, 180],
    [300, 150, 400],
    [220, 350, 280],
    [500, 450, 600]
])
highest_rev = np.argmax(revenue,axis=1)
print(highest_rev)
```

Q3/10 — np.argsort() + Product IDs
A company records revenue for 5 products.
revenue = np.array([850, 400, 950, 600, 750])
product_ids = np.array([301, 302, 303, 304, 305])
Task: Find the product IDs in descending order of revenue.

```python
revenue = np.array([850, 400, 950, 600, 750])
product_ids = np.array([301, 302, 303, 304, 305])
sorted_revenue = np.argsort(revenue)[::-1]
sorted_ids = product_ids[sorted_revenue]
print(sorted_ids)
```

Q4/10 — np.unique() + Frequency
An e-commerce company records customer IDs for purchases.
customer_ids = np.array([
    501, 502, 503, 501, 504,
    502, 501, 505, 503, 502,
    501, 504, 502
])
Task: Find:
1. All unique customer IDs.
2. The number of purchases made by each customer.

```python
customer_ids = np.array([
    501, 502, 503, 501, 504,
    502, 501, 505, 503, 502,
    501, 504, 502
])
unique_ids,counts = np.unique(customer_ids,return_counts=True)
print(unique_ids,counts)
```

Q5/10 — Boolean Masking + Ties
Using the same unique_ids and counts arrays:

Task: Find all customer IDs who made the maximum number of purchases

```python
max_counts = np.max(counts)
mask = counts == max_counts
top_customer = unique_ids[mask]
print(top_customer)
```

Q6/10 — NaN Handling
A company records daily sales for 3 products.
sales = np.array([[100, np.nan, 300],[200, 250, np.nan],[np.nan, 350, 400],[150, 300, 500]])
Task: Count the missing values for each product (column-wise).

```python
sales = np.array([
    [100, np.nan, 300],
    [200, 250, np.nan],
    [np.nan, 350, 400],
    [150, 300, 500]
])
missing_count = np.isnan(sales).sum(axis=0)
print(missing_count)
```

Q8/10 — np.where() — Missing Value Replacement
A company has this inventory data:
stock = np.array([50, np.nan, 30, np.nan, 80, 20])
Task: Replace all missing values with -1 using np.where()

```python
stock = np.array([50, np.nan, 30, np.nan, 80, 20])
clean_data = np.where(np.isnan(stock),-1,stock)
print(clean_data)
```

Q10/10 — Final Business Challenge
A company tracks quarterly revenue for four products.
revenue = np.array([
    [120, 150, 180],
    [200, 170, 230],
    [140, 190, 160],
    [250, 220, 270]
])

product_ids = np.array([101, 102, 103, 104])
Tasks:
1. Calculate total revenue for each product.
2. Find the index of the product with the highest total revenue.
3. Retrieve its product ID and total revenue.

```python
revenue = np.array([
    [120, 150, 180],
    [200, 170, 230],
    [140, 190, 160],
    [250, 220, 270]
])

product_ids = np.array([101, 102, 103, 104])
total_revenue = revenue.sum(axis=1)
idx = np.argmax(total_revenue)
product_id = product_ids[idx]
highest_revenue = total_revenue[idx]
print("Highest Revenue :",highest_revenue,"Product ID :",product_id)
```
