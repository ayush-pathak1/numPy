Q1/15 — 1D Indexing
Given:
import numpy as np
np.array([10, 20, 30, 40, 50, 60])

Print:
1. The first element
2. The last element
3. The element at index 3
Write the code yourself.

```python
import numpy as np
a = np.array([10, 20, 30, 40, 50, 60])
print(a[0])
print(a[-1])
print(a[3])
```

Q2 — 1D Slicing
Given the same array:
a = np.array([10, 20, 30, 40, 50, 60])
Print:
1. Elements from index 1 to 4
2. The first 3 elements
3. The last 2 elements
Write the code without a hint.

```python
a = np.array([10, 20, 30, 40, 50, 60])
print(a[1:5])
print(a[:3])
print(a[-2:])
```

Q4 — 2D Slicing
Given:
a = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
Print:
1. The first 2 rows
2. The last 2 columns
3. The first 2 rows and last 2 columns
Write all three using slicing.

```python
a = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
print(a[:2])
print(a[:,-2:])
print(a[:2,-2:])
```
