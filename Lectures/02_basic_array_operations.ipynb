1. Indexing

Access an element using its index.
a[0]

For 2D arrays:
a[row, column]

```python
import numpy as np
a = np.array([5,6,7])
a[0]
```

2. Array Properties
a.ndim       # Number of dimensions
a.shape      # Rows and columns
a.size       # Total number of elements
a.dtype      # Data type of elements
a.itemsize   # Memory used by each element (bytes)

```python
a = np.array([[1,2],[3,4],[5,6]])
a.ndim
```

```python
a.itemsize
```

```python
a.dtype
```

3. Data Type

Specify the data type while creating an array:
np.array([[1, 2], [3, 4]], dtype=np.float32)
Other example:
dtype=complex

```python
a = np.array([[1,2],[3,4],[5,6]],dtype=np.float32)
a
```

```python
a.itemsize
```

```python
a.size
```

```python
a.shape
```

```python
a = np.array([[1,2],[3,4],[5,6]],dtype=complex)
a
```

4. Creating Arrays
Zeros
np.zeros((3, 4))

Creates an array filled with 0.

```python
np.zeros((3,4))
```

Ones
np.ones((3, 4))

Creates an array filled with 1.

```python
np.ones((3,4))
```

arange()
np.arange(1, 5)
np.arange(1, 10, 2)

Creates values in a specified range.

```python
np.arange(1,5)
```

```python
np.arange(1,10,2)
```

linspace()
np.linspace(1, 5, 10)

Creates 10 evenly spaced values between 1 and 5.

```python
np.linspace(1,5,10)
```

5. Reshaping
a.reshape(2, 3)

Changes the shape without changing the data.

a.reshape(6, 1)

The total number of elements must remain the same.

ravel()
a.ravel()

Flattens the array into a 1D array.

```python
a = np.array([[1,2],[3,4],[5,6]])
a.reshape(2,3)
```

```python
a.reshape(6,1)
```

```python
a.ravel()
```

6. Basic Aggregation
a.min()
a.max()
a.sum()
        min() → minimum value
        max() → maximum value
        sum() → sum of all values

```python
a = np.array([[1,2],[3,4],[5,6]])
```

```python
a.min()
```

```python
a.max()
```

```python
a.sum()
```

7. Axis
a.sum(axis=0)

Calculates the sum column-wise.

a.sum(axis=1)

Calculates the sum row-wise.

axis=0 → down the rows / column-wise
axis=1 → across the columns / row-wise

```python
a.sum(axis=0)
```

```python
a.sum(axis=1)
```

8. Mathematical Functions
np.sqrt(a)

Calculates the square root of each element.

np.std(a)

Calculates the standard deviation.

```python
np.sqrt(a)
```

```python
np.std(a)
```

9. Array Operations
a + b
a - b
a * b
a / b

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6], [7,8]])
a+b
```

```python
a/b
```

```python
a*b
```
