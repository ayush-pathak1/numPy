```python
import numpy as np
```

```python
a = np.arange(12).reshape(3,4)
a
```

1. Iterating Through Rows
A for loop can iterate through a 2D array row by row.
Each iteration gives one complete row.

```python
for row in a:
    print(row)
```

2. Nested Loop
An outer loop can iterate through rows.
An inner loop can iterate through each individual element.

```python
for row in a:
    for cell in row:
     print(cell)
```

3. flatten()
flatten() converts a multi-dimensional array into a 1D array.
Useful when you want to iterate through all elements one by one.

```python
for cell in a.flatten():
    print(cell)
```

4. np.nditer()
np.nditer() is used to iterate through NumPy arrays efficiently.
It provides more control over how the array is traversed.

5. C Order
order='C' → elements are traversed row by row.
This is the normal C-style order.

```python
for x in np.nditer(a,order='C'):
    print(x)
```

6. F Order
order='F' → elements are traversed column by column.
F means Fortran-style order.

```python
for x in np.nditer(a, order='F'):
    print(x)
```

7. external_loop
external_loop makes nditer() return elements in groups/chunks instead of one at a time.

```python
for x in np.nditer(a,order='F',flags=['external_loop']):
    print(x)
```

8. op_flags=['readwrite']
readwrite allows the elements of the original array to be modified during iteration.
Without readwrite, elements are normally read-only.

```python
for x in np.nditer(a,op_flags=['readwrite']):
    x[...]=x*x
```

9. Iterating Two Arrays
np.nditer() can iterate through two arrays at the same time.
If their shapes are compatible, NumPy uses broadcasting.

```python
for x,y in np.nditer([a,b]):
    print(x,y)
```

Key Points
for → simple iteration
flatten() → converts to 1D
nditer() → controlled iteration
C order → row-wise
F order → column-wise
external_loop → groups of elements
readwrite → allows modification
Multiple arrays → can use broadcasting
