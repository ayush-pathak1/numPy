1. 1D Indexing & Slicing
a[0:2]
   Selects elements from index 0 up to, but not including, index 2.
a[-1]
  -1 selects the last element.

```python
import numpy as np
a = np.array([6,7,8])
a[0:2]
```

```python
a[-1]
```

2. 2D Indexing & Slicing
a[1,0]
    Selects an element using [row, column].
a[0:2,2]
    Selects rows 0–1 from column 2.
a[-1]
    Selects the last row.
a[-1,0:2]
    Selects the last row and columns 0–1.
a[:,1:3]
    : means all rows.
    Selects columns 1–2.

```python
a = np.array([[6,7,8],[1,2,3],[9,3,2]])
a[1,0]
```

```python
a[0:2,2]
```

```python
a[-1]
```

```python
a[-1,0:2]
```

```python
a[:,1:3]
```

5. Stacking Arrays
a = np.arange(6).reshape(3,2)
b = np.arange(6,12).reshape(3,2)
    Creates two 3 × 2 arrays.
np.vstack((a,b))
    Stacks arrays vertically (row-wise).
np.hstack((a,b))
    Stacks arrays horizontally (column-wise).

```python
a=np.arange(6).reshape(3,2)
b = np.arange(6,12).reshape(3,2)
```

```python
a
```

```python
b
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

6. Horizontal Splitting
a = np.arange(30).reshape(2,15)
    Creates a 2 × 15 array.
np.hsplit(a,3)
    Splits the array horizontally into 3 parts.
result[0]
result[1]
result[2]
Accesses each split part.

```python
a=np.arange(30).reshape(2,15)
a
```

```python
result = np.hsplit(a,3)
result
```

```python
result[0]
```

```python
result[1]
```

```python
result[2]
```

8. Boolean Indexing
a = np.arange(12).reshape(3,4)
    Creates a 3 × 4 array.
b = a > 4
    Creates a Boolean array.
    Each element becomes True if it is greater than 4, otherwise False.
a[b]
    Selects only elements where the condition is True.
a[b] = -1
    Replaces all elements satisfying the condition with -1.

```python
a = np.arange(12).reshape(3,4)
a
```

```python
b = a > 4
b
```

```python
a[b]
```

```python
a[b]=-1
a
```
