2. Speed Comparison
time.perf_counter()
Used to measure execution time accurately.

Python List:

result = [x + y for x, y in zip(l1, l2)]
Adds elements using a Python loop.

NumPy:

result = a1 + a2
Adds elements directly using NumPy's vectorized operation.
time_taken * 1000
Converts seconds → milliseconds.

```python
SIZE = 1000000

l1 = list(range(SIZE))
l2 = list(range(SIZE))
start = time.perf_counter()
result = [x + y for x, y in zip(l1, l2)]
python_time = time.perf_counter() - start
print("Python list took:", python_time * 1000, "ms")
```

```python
a1 = np.arange(SIZE)
a2 = np.arange(SIZE)
start = time.perf_counter()
result = a1 + a2
numpy_time = time.perf_counter() - start
print("NumPy took:", numpy_time * 1000, "ms")
```
